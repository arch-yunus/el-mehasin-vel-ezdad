# Sentiment Polarity & Dialectic Balance Scoring

Bu not defteri, `Kitâbü'l-Mehâsin ve'l-Ezdâd` korpusundaki metinlerin kutupsallık (polarity), his analizi ve tez-antitez dengesini inceler.

```python
import json
from pathlib import Path

data_path = Path('../data/dialectic_pairs.jsonl')
with open(data_path, 'r', encoding='utf-8') as f:
    pairs = [json.loads(l) for l in f if l.strip()]

# Kutupsallık ve Sentez Kalite Skoru
scored_entries = []
for p in pairs:
    t_len = len(p['thesis']['core_premise'])
    a_len = len(p['antithesis']['core_premise'])
    s_len = len(p['dialectic_synthesis'])
    balance_score = 1.0 - abs(t_len - a_len) / max(t_len, a_len)
    scored_entries.append({
        'id': p['entry_id'],
        'topic': p['topic_slug'],
        'balance_score': round(balance_score, 3),
        'synthesis_length': s_len
    })

print(f"En Dengeli 5 Diyalektik Çift:")
for s in sorted(scored_entries, key=lambda x: -x['balance_score'])[:5]:
    print(f"- {s['id']}: Denge Skoru = {s['balance_score']} | Sentez Uzunluğu = {s['synthesis_length']}")
```
