# Sentiment Polarity & Dialectic Balance in Adab Literature

Bu not defteri, ovgu (el-mehasin) ve yergi (el-mesavi) kutuplarindaki retorik argumanlarin duygu ve ikna kutuplulugunu inceler.

```python
import json

with open('../data/dialectic_pairs.jsonl', 'r', encoding='utf-8') as f:
    data = [json.loads(line) for line in f if line.strip()]

print(f'Analiz edilecek kutup sayisi: {len(data) * 2}')
for item in data:
    print(f"{item['topic_slug']} -> [+] {item['thesis']['concept']} | [-] {item['antithesis']['concept']}")
```
