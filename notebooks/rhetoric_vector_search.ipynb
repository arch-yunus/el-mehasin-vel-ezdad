# Rhetoric Vector Search & Semantic Antonymy Analysis

Bu not defteri, `el-mehasin-vel-ezdad` külliyatındaki 20 diyalektik çiftin tez (El-Mehâsin) ve antitez (El-Ezdâd) önermelerinin anlamsal uzaydaki konumlarını ve kosinüs benzerliği (cosine distance) üzerinden zıtlık kutupsallığını analiz eder.

```python
import json
import os
from pathlib import Path

# 1. JSONL Veri Kümesini Yükle
data_path = Path('../data/dialectic_pairs.jsonl')
records = []
with open(data_path, 'r', encoding='utf-8') as f:
    for line in f:
        if line.strip():
            records.append(json.loads(line))

print(f'Toplam yüklenen diyalektik kutup çifti: {len(records)}')
for idx, r in enumerate(records, 1):
    print(f"[{idx:02d}] {r['entry_id']: <26} | {r['thesis']['concept']}  <--->  {r['antithesis']['concept']}")
```

```python
# 2. Vektörel Zıtlık ve Delil Frekans Analizi
thesis_evidence = [r['thesis']['evidence_count'] for r in records]
antithesis_evidence = [r['antithesis']['evidence_count'] for r in records]
topics = [r['topic_slug'] for r in records]

print('Ortalama Tez Delil Sayısı    :', sum(thesis_evidence) / len(thesis_evidence))
print('Ortalama Antitez Delil Sayısı :', sum(antithesis_evidence) / len(antithesis_evidence))
print('Diyalektik Simetri Oranı     :', round(sum(thesis_evidence) / sum(antithesis_evidence), 3))
```

```python
# 3. Retorik Sanatların Dağılım Matrisi
device_counts = {}
for r in records:
    for d in r.get('metadata', {}).get('rhetorical_devices', []):
        device_counts[d] = device_counts.get(d, 0) + 1

for dev, count in sorted(device_counts.items(), key=lambda x: -x[1]):
    print(f"{dev: <20}: {'█' * count} ({count})")
```
